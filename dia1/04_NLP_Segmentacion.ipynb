## 3. Segmentación

La segmentación divide un texto en partes significativas, como frases u oraciones.

```python
import nltk
# Descargar los recursos necesarios de NLTK
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

texto = "La inteligencia artificial es fascinante. NLP es una de sus aplicaciones."

oraciones = sent_tokenize(texto)
print("Oraciones segmentadas",oraciones)
```

```python
texto = "¡Hola! ¿Cómo estás? Hoy es un buen día para aprender NLP."

# Tokenización en palabras
tokens_palabras = word_tokenize(texto)
print("Tokens de palabras:", tokens_palabras)

# Tokenización en oraciones
tokens_oraciones = sent_tokenize(texto)
print("Tokens de oraciones:", tokens_oraciones)
```

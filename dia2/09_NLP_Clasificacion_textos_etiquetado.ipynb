# ETIQUETADO EN ESPAÑOL

```python
nltk.download('cess_esp')
```

```python
from nltk.corpus import cess_esp as cess
from nltk import UnigramTagger as ut
from nltk import BigramTagger as bt
```

# Entrenamiendo del tagger por unigramas

```python
cess_sents = cess.tagged_sents()
fraction = int(len(cess_sents)*90/100)
cess_sents = cess.tagged_sents()
uni_tagger = ut(cess_sents[:fraction])
uni_tagger.evaluate(cess_sents[fraction:])
```

```python
uni_tagger.tag("Yo soy una persona muy amable".split(" "))
```

```python
def interpretar_etiqueta_cess(etiqueta):
    explicacion = {
        'n': 'sustantivo',
        'p': 'pronombre',
        'v': 'verbo',
        'a': 'adjetivo',
        'd': 'determinante',
        'r': 'adverbio',
        'c': 'conjunción',
        's': 'preposición',
        'i': 'interjección',
        'f': 'signo de puntuación',
        'z': 'número',
        # Puedes extender esto para más detalle
    }

    if etiqueta[0] in explicacion:
        print(f"{etiqueta}: {explicacion[etiqueta[0]]}")
    else:
        print(f"{etiqueta}: etiqueta desconocida (o no implementada)")

interpretar_etiqueta_cess('vsip1s0')
```

# Entrenamiento del tagger por bigramas

```python
fraction = int(len(cess_sents)*90/100)
bi_tagger = bt(cess_sents[:fraction])
bi_tagger.evaluate(cess_sents[fraction:])
```

```python
bi_tagger.tag("Yo soy una persona muy amable".split(" "))
```

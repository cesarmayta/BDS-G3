```python
import tensorflow as tf
from tensorflow import keras
```

# CARGAMOS DATASET IMDB

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data()
# Obtener el índice de palabras
word_index = keras.datasets.imdb.get_word_index()
reverse_word_index = {value: key for key, value in word_index.items()}

# Tomar un ejemplo del dataset
sample_review = x_train[0]

# Convertir los índices en palabras
decoded_review = " ".join([reverse_word_index.get(i - 3, "?") for i in sample_review])

print("Reseña decodificada:")
print(decoded_review)
```

```python
vocab_size = 10000
max_length = 100

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=vocab_size)
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=max_length, padding='post')
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=max_length, padding='post')
```

# CREAMOS RED NEURONAL RECURRENTE

```python
model_rnn = keras.Sequential([
    keras.layers.Embedding(input_dim=vocab_size, output_dim=128, input_length=max_length),
    keras.layers.SimpleRNN(64),  # 🔹 Neurona recurrente simple
    keras.layers.Dense(1, activation='sigmoid')  # Salida binaria
])
model_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

# ENTRENAMOS LA RED RECURRENTE

```python
model_rnn.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))
```

# EVALUAMOS LA RED NEURONAL

```python
test_loss, test_acc = model_rnn.evaluate(x_test, y_test)
print(f"\n🔹 Precisión del modelo RNN: {test_acc:.4f}")
```

# CREAMOS RNN CON LSTM

```python
# 1️⃣ Modelo con LSTM
model_lstm = keras.Sequential([
    keras.layers.Embedding(input_dim=vocab_size, output_dim=128, input_length=max_length),
    keras.layers.LSTM(64),  # 🔹 Neurona LSTM
    keras.layers.Dense(1, activation='sigmoid')  # Salida binaria
])

# 2️⃣ Compilar y entrenar
model_lstm.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_lstm.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))

# 3️⃣ Evaluar modelo
test_loss, test_acc = model_lstm.evaluate(x_test, y_test)
print(f"\n🔹 Precisión del modelo LSTM: {test_acc:.4f}")
```

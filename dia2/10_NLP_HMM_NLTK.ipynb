# ENTRENAMIENTO DE UN MODELO MARKOVIANO LATENTE(HMM) INGLES

# ETIQUETADO EN INGLES CON NLTK

```python
import nltk
nltk.download('treebank')
from nltk.corpus import treebank
```

# CARGAMOS EL DATASET

```python
train_data = treebank.tagged_sents()[:3900]
train_data
```

# CONSTRUIMOS UN MODELO HMM CON NLTK

```python
from nltk.tag import hmm
tagger = hmm.HiddenMarkovModelTrainer().train_supervised(train_data)
tagger
```

# PROBAMOS EL MODELO

```python
tagger.tag('Pierre Vinken will get old'.split())
```

```python
nltk.download('tagsets_json')
for tag in ['NNP', 'MD', 'VB', 'JJ']:
  print(nltk.help.upenn_tagset(tag))
```

```python
tagger.evaluate(treebank.tagged_sents()[:2900])
```

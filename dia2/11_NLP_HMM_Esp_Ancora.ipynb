# TRANSFORMACIÓN DEL CORPUS

```python
data_file = open("UD_Spanish-AnCora/es_ancora-ud-dev.conllu", "r", encoding="utf-8")
word_list = []
for tokenlist in parse_incr(data_file):
  words = []
  for token in tokenlist:
    tag = token['upos']
    pal = token['form']
    words.append((pal,tag))
  word_list.append(words)

len(word_list)
```

# DIVIDIMOS DATA EN TRAIN Y TEST

```python
from sklearn.model_selection import train_test_split

data_train,data_test = train_test_split(word_list,test_size=0.30,random_state=42)
```

# CREAMOS EL MODELO HMM

```python
import nltk
from nltk.tag import hmm

tagger = hmm.HiddenMarkovModelTrainer().train_supervised(data_train)
```

# PROBAMOS EL MODELO

```python
tagger.tag('el mundo es pequeño'.split())
```

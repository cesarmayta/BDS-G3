```python
import stanza
stanza.download('es')
```

# CREANDO UN PIPELINE

```python
nlp = stanza.Pipeline('es',processors='tokenize,pos')
doc = nlp('yo soy una persona muy amable')
```

```python
doc
```

```python
for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text,word.pos)
```

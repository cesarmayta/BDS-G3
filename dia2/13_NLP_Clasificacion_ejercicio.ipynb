```python
DATASET_PATH = path + '/vacunasXFrase.csv'
import pandas as pd

df = pd.read_csv(DATASET_PATH,sep='|')
df.head(5)
```

```python
df_tweets = df['tweet'].head(10).copy()
df_tweets
```

```python
import stanza
stanza.download('es')
```

```python
nlp = stanza.Pipeline('es')
```

```python
def tag_tweet(tweet):
  doc = nlp(tweet)
  tags = [(word.text,word.upos) for sent in doc.sentences for word in sent.words]
  return tags
```

```python
tags = tag_tweet('Que informe @FernanQuirosBA cuándo van a dar')
tags
```

```python
df_tweets['tags'] = df['tweet'].apply(tag_tweet)
df_tweets
```

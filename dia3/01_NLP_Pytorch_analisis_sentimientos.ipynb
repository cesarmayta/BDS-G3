# IMPORTAMOS LIBRERIAS NECESSARIOS PARA NLP

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchtext
import tqdm
```

# PASO 1 - TOKENIZACIÓN

```python
tokenizer = torchtext.data.utils.get_tokenizer('basic_english')
texto = "Hello World! How Are you today?"
tokenizer(texto)
```

```python
def tokenize_example(example,tokenizer,max_length):
  tokens = tokenizer(example["text"][:max_length])
  return {"tokens":tokens}
```

```python
MAX_LENGTH = 256
train_data = train_data.map(
    tokenize_example,fn_kwargs={"tokenizer":tokenizer,"max_length":MAX_LENGTH}
)
train_data
```

```python
test_data = test_data.map(
    tokenize_example,fn_kwargs={"tokenizer":tokenizer,"max_length":MAX_LENGTH}
)
test_data
```

```python
train_data[0]["tokens"][:10]
```

# PASO 2 - DIVIDIEMOS NUESTRO DATASET EN TRAIN Y VALIDATE

```python
TEST_SIZE = 0.25
train_valid_data = train_data.train_test_split(test_size=TEST_SIZE)
train_data = train_valid_data['train']
valid_data = train_valid_data['test']

print(f' TRAIN DATA : {len(train_data)}')
print(f' VALIDATE DATA : {len(valid_data)}')
print(f' TEST DATA : {len(test_data)}')
```

# CREAMOS UN VOCABULARIO DE PALABRAS

```python
MIN_FREQ = 5
special_tokens = ["<unk>","<pad>"]

vocab = torchtext.vocab.build_vocab_from_iterator(
    train_data["tokens"],
    min_freq=MIN_FREQ,
    specials=special_tokens
)
```

```python
len(vocab)
```

```python
vocab["the"]
```

```python
unk_index = vocab["<unk>"]
pad_index = vocab["<pad>"]
print(f'indice unk : {unk_index}')
print(f'indice pad : {pad_index}')
```

```python
vocab.set_default_index(unk_index)
```

## NUMERIZAR DATA

```python
test_data
```

```python
def num_example(example,vocab):
  ids = vocab.lookup_indices(example['tokens'])
  return {'ids':ids}
```

```python
train_data = train_data.map(num_example,fn_kwargs={'vocab':vocab})
valid_data = valid_data.map(num_example,fn_kwargs={'vocab':vocab})
test_data = test_data.map(num_example,fn_kwargs={'vocab':vocab})
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Función para generar el Positional Encoding
def positional_encoding(seq_len, d_model):
    """
    Genera el Positional Encoding para una secuencia de longitud `seq_len` y dimensión `d_model`.

    Args:
    seq_len (int): Longitud de la secuencia.
    d_model (int): Dimensión del espacio de representación.

    Returns:
    np.ndarray: Positional Encoding de tamaño (seq_len, d_model)
    """
    # Creamos un array de posiciones de 0 hasta seq_len-1
    position = np.arange(seq_len)[:, np.newaxis]

    # Calculamos los divisores según la fórmula del paper original de "Attention is All You Need"
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))

    # Calculamos el Positional Encoding
    pos_enc = np.zeros((seq_len, d_model))
    pos_enc[:, 0::2] = np.sin(position * div_term)  # Aplicamos la función seno en las posiciones pares
    pos_enc[:, 1::2] = np.cos(position * div_term)  # Aplicamos la función coseno en las posiciones impares

    return pos_enc
```

```python
# Parámetros
seq_len = 50  # Longitud de la secuencia
d_model = 128  # Dimensión de los vectores de la secuencia

# Generar Positional Encoding
pos_enc = positional_encoding(seq_len, d_model)

# Visualización de los primeros 20 valores
plt.figure(figsize=(10, 8))
plt.imshow(pos_enc[:20, :], cmap='viridis', aspect='auto')
plt.colorbar()
plt.title('Visualización del Positional Encoding para una secuencia')
plt.xlabel('Dimensión del vector')
plt.ylabel('Posición de la secuencia')
plt.show()
```

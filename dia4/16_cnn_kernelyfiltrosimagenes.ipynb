# EJEMPLO DE USO DEL KERNEL(FILTRO) EN IMAGENES

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage as nd
from skimage import io,color
```

# CARGAMOS LA FOTO DE EJEMPLO

```python
img = io.imread('/content/foto.jpg')
plt.imshow(img)
```

# CONVERTIR IMAGEN

```python
img_gray = color.rgb2gray(img)
plt.imshow(img_gray)
```

# CREAMOS UN KERNAL(FILTRO)

```python
kernel1 = np.array([[-1,0,-1],
                   [-1,0,-1],
                   [-1,0,-1]])
kernel1
```

# APLICAMOS CONVOLUCIÓN USANDO EL KERNEL

```python
img_new = nd.convolve(img_gray,kernel1)
plt.imshow(img_new)
```

# APLICAMOS OTROS FILTROS

```python
kernel2 = np.array([[ 0, 0, 0],
                   [ 0, 1, 0],
                   [ 0, 0, 0]])
img_new2 = nd.convolve(img_gray,kernel2)
plt.imshow(img_new2)
```

```python
kernel_blur = np.array([[ 0.0625, 0.125, 0.0625],
                   [ 0.125, 0.25, 0.125],
                   [ 0.0625, 0.125, 0.0625]])
img_new3 = nd.convolve(img_gray,kernel_blur)
plt.imshow(img_new3)
```

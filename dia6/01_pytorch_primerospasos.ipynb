# CREAMOS NUESTRA PRIMERA RED NEURONAL CON PYTORCH

## 1 - IMPORTAMOS LIBRERIAS

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

## 2 - CREAMOS UNA CLASE PARA NUESTRO MODELO DE RED NEURONAL

```python
class SimpleNN(nn.Module):
  def __init__(self):
    super(SimpleNN,self).__init__()
    self.fc = nn.Linear(2,1) #creamos una capa con 2 neuronas de entrara y 1 de salida

  def forward(self,x):
    return self.fc(x)
```

## 3 - CREAMOS UN OBJETO DE CLASE DE NUESTRO MODELO Y DEFINIMOS FUNCIÓN DE PERDIA Y OPTIMIZADOR

```python
model = SimpleNN()
criterion = nn.MSELoss()# error cuadratico medio
optimizer = optim.SGD(model.parameters(),lr=0.01) # Descenso de Gradiente Estocástico (SGD, Stochastic Gradient Descent),
```

## 4 - CREAMOS UN DATASET DE PRUEBA PARA ENTRENAR NUESTRO MODELO

```python
inputs = torch.tensor([[1.0,2.0],[2.0,3.0]],dtype=torch.float32)
inputs
```

```python
labels = torch.tensor([[1.0],[2.0]],dtype=torch.float32)
labels
```

## 5 - ENTRENAMOS EL MODELO

```python
epochs = 10
for epoch in range(epochs):
  outputs = model(inputs) #hacemos un predicción con el modelo
  loss = criterion(outputs,labels) # calculamos la perdida

  #backprogration y optimización
  optimizer.zero_grad() # reiniciar los gradientes para evitar acumulaciones incorrectas.
  loss.backward() # Calcular los gradientes
  optimizer.step() # Actualizar los pesos del modelo
  print(f'epoca {epoch + 1}, loss: {loss.item():.4f}')

  if(epoch + 1) % 10 == 0:
    print(f'resultado final : epoca {epoch + 1}, loss : {loss.item():.4f}')
```

## 6 PREDICCIONES DEL MODELO

```python
import numpy as np
print(f'valores predichos : {np.round(model(inputs).detach().numpy())}')
print(f'valores reales :{labels}')
```
